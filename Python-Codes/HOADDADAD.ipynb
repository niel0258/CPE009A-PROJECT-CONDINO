```python
import os

#create folder
fol_pat = 'calculations'

if not os.path.exists(fol_pat):
    os.mkdir(fol_pat)

#create file
init_dir = os.path.join(fol_pat,'__init__.py')
if not os.path.isfile(init_dir):
    init_fi = open(init_dir,"x")
    init_fi.close()

g_dir = os.path.join(fol_pat,'geometry.py')

if not os.path.isfile(g_dir):
    g_fi = open(g_dir,"w")
    g_fi.write('def sq_area(s):\n	return s**2')
    g_fi.close()
```

```python
from calculations import geometry

print(geometry.sq_area(4))
```
