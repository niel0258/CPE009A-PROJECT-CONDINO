```python
import pandas as pd
#creates a data frame object
f_df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "Age": [25, 30, 35, 28, 40],
    "Score": [88.5, 92.0, 79.5, 85.0, 90.5]
})#example data frame
print(f_df)
f_df.info()
f_df.describe()#gets different measurements for data like count,mean,etc.
```

```python
import pandas as pd
import random as rand

df = pd.DataFrame(
    [[rand.randint(1,10) for _ in range(2)] for _ in range(7)],#data
    index=pd.date_range(start="2-12-26",periods=7),
    columns=['Collected bills','Collected pennies']
)

df
```

        Created by Wes McKinney on 2008 out of the need for a high performance, flexible tool to perform quantitative analysis on financial data, The pandas module is a very helpful module for those who do data analysis and manipulation. It is a fast and efficient module for data analytics and widely used among fields. It can do many things related to data science like finding the mean,standard deviation,variance,min and max value,etc. of a given data.

The methods of data frame that I used are the info() and describe() methods. The info() method returns information about the data frame, e.g.: its type, number of columns, the type of the data in the column,number of entries, and the memory size in bytes. The describe() method is used to get measure of central tendencies like the mean,percentiles,standard deviation, min and max, count of the data in each columns. The only function I used is date_range() which returns a list of dates starting from the value of the start parameter and number of days given by the periods parameter.
