```python
#Leap year or Common Year checker

year = int(input("Enter a year: "))
messages = ["Not within the Gregorian Calendar","Common year","Leap year"]

#if the year number isn't divisible by four, it's a common year;
#otherwise, if the year number isn't divisible by 100, it's a leap year;
#otherwise, if the year number isn't divisible by 400, it's a common year;
#otherwise, it's a leap year.

if year < 1582:
  print(messages[0])
elif (year % 4) > 0: #not divisible by 4
  print(messages[1])
elif (year % 100) > 0:
  print(messages[2])
elif (year % 400) > 0:
  print(messages[1])
else:
  print(messages[2])
```

```python
#Block pyramid height finder
height = 0
base = 0 #Needed no. of box for the base of the pyramid
blocks = int(input("Input number of blocks: "))

while blocks > 0:
  base += 1
  for block in range(1,base + 1): #base +1 for expected behaviour e.g if base = 6, loop 6 times not 5
    blocks -= 1
    if blocks >= 0 and block == base: #check if completed the base
      height += 1
    elif blocks < 0:#not counted if not completed the base
      break

print("The height of the pyramid:",height)
```

```python
#Remove duplicate numbers (loop approach)
my_list = [1, 2, 4, 4, 1, 4, 2, 6, 2, 9]
seen_elem = []
dup_index = []

for i,elem in enumerate(my_list):
  if elem in seen_elem:
    dup_index.append(i)
  else:
    seen_elem.append(elem)

for i in reversed(dup_index):#pop from the highest index in order to avoid index shifting
  my_list.pop(i)

print("The list with unique elements only:")
print(my_list)
```

```python
#Remove duplicate numbers (set approach)
my_list = [1, 2, 4, 4, 1, 4, 2, 6, 2, 9]

my_list = list(set(my_list))#Turn to set to remove duplicates

print("The list with unique elements only:")
print(my_list)
```

```python
#Student Score Evaluator

#inputted scores
scores = [85, 72, 90, 60, 78, 55, 88]
bonus_flags = [1, 0, 1, 0, 1, 0, 1]

#in code lists
passed_num = 0
failed_num = 0
certificate_num = 0
results = []

for student_num,score in enumerate(scores):#enumerate() to also get the index
  passed = bool(score >= 75)
  remark = ""
  has_bonus = bool(bonus_flags[student_num] & 1)

  if passed:
    remark = "PASSED"
    passed_num += 1
    if has_bonus:
      certificate_num += 1
  else:
    remark = "FAILED"
    failed_num += 1

  results.append("STUDENT " + str(student_num+1) + ": " + remark)


print("Total Student: ", len(scores))
print("Passed: ",passed_num)
print("Failed: ",failed_num)
print("Certificates Earned:",certificate_num)
```

```python
#Bubble Sort

score_list = [22,24,27,39,55,53,60,41,33]
swapped = True #To Enter the bubble sort loop

while swapped:
    swapped = False
    for i in range(len(score_list) - 1):
        if score_list[i] > score_list[i + 1]:
            swapped = True
            score_list[i], score_list[i + 1] = score_list[i + 1], score_list[i]

print(score_list)
```
