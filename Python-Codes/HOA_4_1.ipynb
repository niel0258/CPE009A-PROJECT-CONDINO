```python
def is_year_leap(year) -> bool:
  if (year % 4) > 0: #Common
    return False
  elif (year % 100) > 0: #Leap
    return True
  elif (year % 400) > 0:#Commom
    return False
  else:#Leap
    return True

def days_in_month(year, month) -> int:
  day_list = [31,28,31,30,31,30,31,31,30,31,30,31]
  if month > len(day_list): return #return none if month does not makes sense
  if (month == 2) and (is_year_leap(year)):
    day_list[1] = 29
  return day_list[month - 1]

def day_of_year(year, month, day) -> int:
  c_days = 0 # cumulative days
  for month_count in range(month):
    dy_in_mo = days_in_month(year,month_count+1)
    if month_count == month -1: #reached the month given in the argument
      if day > dy_in_mo: #Given day is greater than num of days month in that month
        return #The argument does not make sense
      else:
        c_days += day
    else:
      c_days += dy_in_mo
  return c_days


#For testing purpose
test_years = [1900, 2000, 2016, 1987]
test_months = [2, 2, 1, 11]
test_days = [20, 29, 20, 30]
test_results = [51, 60, 20, 334]

for i in range(len(test_years)):
  yr = test_years[i]
  mo = test_months[i]
  dy = test_days[i]
  result = day_of_year(yr, mo,dy)
  print(yr, mo, dy, "->", f"({result})",end="")
  if result == test_results[i]:
    print("OK")
  else:
    print("Failed")
```

```python
#1 American mile = 1609.344 metres;
#1 American gallon = 3.785411784 litres.

def liters_100km_to_miles_gallon(liters):
  #convert to mil/gal
  gal = liters/3.785411784
  mil = 100000/1609.344
  return mil/gal

def miles_gallon_to_liters_100km(miles):
  km = (miles * 1609.344) / 100000

  lit = 3.785411784
  return lit/km

print(liters_100km_to_miles_gallon(3.9))
print(liters_100km_to_miles_gallon(7.5))
print(liters_100km_to_miles_gallon(10.))
print(miles_gallon_to_liters_100km(60.3))
print(miles_gallon_to_liters_100km(31.4))
print(miles_gallon_to_liters_100km(23.5))
```

```python
def compute_average(grades):
  c_sum = 0
  grade_sz = len(grades)

  for i in range(grade_sz): #summation of grades
    c_sum += grades[i]

  mean = c_sum / grade_sz #average
  rep_mean = round(mean,2) #round

  return rep_mean

grade = (84.5,87.3,93,90,95)
print(compute_average(grade))
```

```python
def get_price(products,item_name):

  price = products.get(item_name)

  if price is None:# if not found
    return "Item does not exist"
  else:
    return ("Price: " + str(price)) #return price string

products_sold = {
    "gun": 500,
    "cigar": 2,
    "hat":25,
    "boots":37,
}

print(get_price(products_sold,input("Input item name:")))
```

```python
def analyze_sales(sales_data):
  t_sales = 0 #total
  h_sale = 0 # highest sale

  for sale in sales_data.values():
    t_sales += sale #compile all sales
    if h_sale < sale:
      h_sale = sale #get the highest sale

  a_sale = t_sales / len(sales_data) #average sale

  sale_data = {
      "Total Sale":t_sales,
      "Average Sale":a_sale,
      "Highest Sale":h_sale,
  }

  return sale_data #return dictionary

daily_sale = {
    "Monday":200,
    "Tuesday":300,
    "Wednesday":260,
    "Thursday":700,
    "Friday":500,
    "Saturday":1000,
    "Sunday":600
}

sale_det = analyze_sales(daily_sale)

print("Total Sale:",sale_det["Total Sale"])
print("Average Sale:",sale_det["Average Sale"])
print("Highest Sale:",sale_det["Highest Sale"])
```
