<h1>Valid String</h1>

```python
from re import split

dictionary = ["code","wars"]
test_cases = ["codewars","codewar","cdewars","warcode"]

def create_delim(matched:list) -> str:
    delim = ''

    for matched_word in matched:
        delim += matched_word + "|"

    return delim

def valid_str(seq:list, word:str)-> dict:
    dp = [False] * (len(word) + 1) 
    dp[len(word)] = True #Base Case
    
    words = {
        "result" : False,
        "matched" : [],
        "unmatched" : []
    }
    
    for i in range(len(word)-1,-1,-1):
        for w in seq:
            if not (i + len(w) <= len(word)): continue #not enough characters to compare
            if word[i:i+len(w)] == w:
                dp[i] = dp[i+len(w)]
                words['matched'].append(word[i:i+len(w)])
            if dp[i]:#is seen
                break
                
    words['result'] = dp[0]
    
    words['unmatched'].append(''.join(split(create_delim(words["matched"]),word)))
    
    return words

for tst_case in test_cases:
    res = valid_str(dictionary,tst_case)
    print(f"{tst_case} => {res['result']} -> match '{res['matched']}' unmatched '{res["unmatched"]}'")
```

<h1>Convert String to camel case</h1>

```python
from re import split

test_cases = ["the-stealth-warrior","The_Stealth_Warrior","The_Stealth-Warrior"]

def camelCase(orig_str:str):
    mod_strs = split('-|_',orig_str)
    for i in range(1,len(mod_strs)):
        mod_strs[i] = mod_strs[i].capitalize()
    
    return ''.join(mod_strs)
#test cases
for case in test_cases:
    print(f"\"{case}\" gets converted to \"{camelCase(case)}\"")
```

<h1>SPLIT STRING</h1>

```python
def split_strs(s:str) -> list:
    split_s = list(s)
    s_len = len(s)
    
    if s_len % 2 > 0:
        split_s.append("_")
    
    for i in range(len(split_s),-1,-1):
        if i%2:
            split_s[i-1] += split_s[i]
            split_s.pop(i)
    
    return split_s

print("abc =>",split_strs('abc'))
print("abcdef =>",split_strs('abcdef'))
```

<h1>Increment String</h1>

```python
foo_case = ["foo","foobar23","foo0042","foo9","foo099"]

def inc_str(word:str,current_index:int) -> str:
    word_list = word if type(word)  == list else list(word)
    
    if word_list[current_index].isdigit():
        if word_list[current_index] == '9':
            word_list[current_index] = '0'
            return inc_str(word_list,current_index - 1)
        else:
            word_list[current_index] = str(int(word_list[current_index]) + 1)
    else:
        word_list.insert(len(word_list) if current_index == -1 else current_index + 1,'1')
    
    return ''.join(word_list)


for foo in foo_case:
    print(inc_str(foo,-1))
```
