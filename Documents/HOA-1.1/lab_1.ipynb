```python
# Task 3:
conv_factor = 1.60934
kilometers = 12.25
miles = 7.38
miles_to_kilometers = miles * conv_factor
kilometers_to_miles = kilometers / conv_factor
print(miles, "miles is", round(miles_to_kilometers, 2), "kilometers")
print(kilometers, "kilometers is", round(kilometers_to_miles, 2), "miles")
```
